# Exploración de datos - Olist
El objetivo de este archivo es entender los datos y su calidad

## Entendimiento de los datos

```python
#Importamos librerias 
import pandas as pd
from pathlib import Path


data_path = Path("../data/raw")

# Identificamos los archivos CSV
files = list(data_path.glob("*.csv"))

for file in files:
    print(file.name)
```

```python
# Cargamos todos los archivos CSV en un diccionario e imprimimos la cantidad de registros 
# que tiene cada archivo, esto nos permitira saber con cuantos datos vamos a trabajar
dataframes = {}

for file in files:
    df = pd.read_csv(file)
    dataframes[file.stem] = df

    print(f"{file.name}: {df.shape[0]:,} filas y {df.shape[1]} columnas")
```

```python
# Revisamos los tipos de datos de cada tabla

for nombre, df in dataframes.items():
    print(f"\n{'=' * 50}")
    print(f"Tabla: {nombre}")
    df.info()
```

```python
# Mostramos los primeros 5 registros de cada tabla, esto nos permite observar su estructura y algunos valores

for nombre, df in dataframes.items():
    print(f"\n{'=' * 50}")
    print(f"Tabla: {nombre}")
    display(df.head())
```

## Calidad de los datos

```python
# Analizamos valores nulos 

# Revisar valores nulos en todas las tablas
for nombre, df in dataframes.items():
    print(f"\n{'=' * 50}")
    print(f"Tabla: {nombre}")

    nulos = df.isnull().sum()
    porcentaje = (nulos / len(df)) * 100

    resumen_nulos = pd.DataFrame({
        "Valores nulos": nulos,
        "Porcentaje (%)": porcentaje.round(2)
    })

    print(resumen_nulos[resumen_nulos["Valores nulos"] > 0])
```

```python
# Revisamos registros completamente duplicados

for nombre, df in dataframes.items():
    duplicados = df.duplicated().sum()

    print(f"\nTabla: {nombre}")
    print(f"Registros duplicados: {duplicados:,}")
```

```python
# Analizar duplicados de geolocalización

geo = dataframes["olist_geolocation_dataset"]

duplicados = geo.duplicated().sum()
total = len(geo)

print(f"Total de registros: {total:,}")
print(f"Registros duplicados: {duplicados:,}")
print(f"Porcentaje de duplicados: {(duplicados / total) * 100:.2f}%")
print(f"Registros restantes sin duplicados exactos: {len(geo.drop_duplicates()):,}")
```

```python
# Revisar códigos postales repetidos

codigos_repetidos = geo["geolocation_zip_code_prefix"].value_counts()

print("Códigos postales repetidos:")
print(codigos_repetidos.head(10))
```

```python
# Analizar cuántas coordenadas distintas tiene cada código postal

geo_por_codigo = (
    geo.groupby("geolocation_zip_code_prefix")
       .agg(
           registros=("geolocation_zip_code_prefix", "size"),
           coordenadas_distintas=("geolocation_lat", "nunique"),
           ciudades=("geolocation_city", "nunique")
       )
       .sort_values("registros", ascending=False)
)

print(geo_por_codigo.head(10))
```

```python
# Revisar identificadores y posibles duplicados por clave

claves = {
    "olist_customers_dataset": ["customer_id"],
    "olist_orders_dataset": ["order_id"],
    "olist_order_items_dataset": ["order_id", "order_item_id"],
    "olist_products_dataset": ["product_id"],
    "olist_sellers_dataset": ["seller_id"],
    "olist_order_reviews_dataset": ["review_id"],
    "product_category_name_translation": ["product_category_name"]
}

for tabla, columnas in claves.items():
    df = dataframes[tabla]

    duplicados = df.duplicated(subset=columnas).sum()

    print(f"\nTabla: {tabla}")
    print(f"Clave revisada: {columnas}")
    print(f"Registros duplicados por clave: {duplicados:,}")
```

```python
# Identificar reseñas con review_id repetido

reviews = dataframes["olist_order_reviews_dataset"]

resumen_reviews = (
    reviews.groupby("review_id")
    .agg(
        cantidad_registros=("review_id", "size"),
        pedidos_distintos=("order_id", "nunique"),
        calificaciones_distintas=("review_score", "nunique")
    )
)

# Mostrar únicamente los identificadores repetidos
reviews_repetidas = resumen_reviews[
    resumen_reviews["cantidad_registros"] > 1
]

print(f"Review ID repetidos: {len(reviews_repetidas):,}")

display(reviews_repetidas.head(10))
```

```python
# Mostrar ejemplos de reseñas con identificadores repetidos

ids_repetidos = reviews_repetidas.index

display(
    reviews[reviews["review_id"].isin(ids_repetidos)]
    .sort_values("review_id")
    .head(20)
)
```

```python
# Comprobar la unicidad de la combinación review_id + order_id

duplicados_review_order = reviews.duplicated(
    subset=["review_id", "order_id"]
).sum()

print(
    "Duplicados en la combinación review_id + order_id:",
    duplicados_review_order
)
```
